```python
# data cleaning

#removing extra spaces
df.columns = df.columns.str.strip()

# making entries common
df['Item Fat Content'] = df['Item Fat Content'].replace('LF','Low Fat')
df['Item Fat Content'] = df['Item Fat Content'].replace('low fat','Low Fat')
df['Item Fat Content'] = df['Item Fat Content'].replace('reg','Regular')
```

```python
### business requirements 

# KPI requirements

#total sales
total_sales = df['Sales'].sum()
print("Total sales:",total_sales)

#average sales
avg_sales = df['Sales'].mean()
print("Average sales:",avg_sales)

#no of items
items_count = df['Item Identifier'].count()
print("No of items:",items_count)

#average ratings 
avg_ratings = df['Rating'].mean()
print("Average ratings:",avg_ratings)


```

```python
# CHART REQUIREMENTS
```

```python

# Total Sales by Fat Content
sales_by_fat = df.groupby('Item Fat Content')['Sales'].sum()

colors = ['#6ABA2E', '#F4C430']  

plt.pie(sales_by_fat, labels=sales_by_fat.index, autopct='%.1f%%',startangle=90,colors=colors)

plt.title('Sales by Fat Content')
plt.axis('equal')
plt.show()
```

```python

# Total Sales by Item Type
sales_by_type = df.groupby('Item Type')['Sales'].sum().sort_values(ascending=False)

plt.figure(figsize=(10,6))

bars = plt.bar(
    sales_by_type.index,
    sales_by_type.values,
    color='#F4C430'
)

plt.xticks(rotation=90)
plt.xlabel('Item Type')
plt.ylabel('Total Sales')
plt.title('Total Sales by Item Type')

for bar in bars:
    plt.text(
        bar.get_x() + bar.get_width()/2,
        bar.get_height(),
        f'{bar.get_height():,.0f}',
        ha='center',          # Center the label
        va='bottom',
        fontsize=8
    )

plt.tight_layout()
plt.show()
```

```python
# Fat content by outlet for total sales

grouped = df.groupby(['Outlet Location Type','Item Fat Content'])['Sales'].sum().unstack()
grouped = grouped[['Regular','Low Fat']]

ax = grouped.plot(kind='bar',color=['#6CB33F', '#F7C600'],
                  figsize=(8,5),
                  title='Outlet Tier by Item Fat Content')
plt.xlabel('Outlet Location Tier')
plt.ylabel('Total Sales')
plt.legend(title='Item Fat Content')
plt.tight_layout()
plt.show()
```

```python
# Total Sales by Outlet Establishment (Line Chart)

sales_by_outlet_es = df.groupby('Outlet Establishment Year')['Sales'].sum().sort_index()

plt.figure(figsize=(10,6))

# line
plt.plot(
    sales_by_outlet_es.index,
    sales_by_outlet_es.values,
    color='#5FAF2D',
    marker='o',
    linewidth=3
)

# shaded area
plt.fill_between(
    sales_by_outlet_es.index,
    sales_by_outlet_es.values,
    color='#F7C600',      
    alpha=0.35            
)

plt.xlabel('Outlet Establishment Year')
plt.ylabel('Total Sales')
plt.title('Total Sales by Outlet Establishment Year')

for x, y in zip(sales_by_outlet_es.index, sales_by_outlet_es.values):
    plt.text(x, y, f'{y:,.0f}', ha='center', va='bottom', fontsize=8)

plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
# Sales by Outlet size

sales_by_Outlet_size = df.groupby('Outlet Size')['Sales'].sum()

colors = ['#6ABA2E', '#F4C430','#90EE90']  

plt.pie(sales_by_Outlet_size, labels=sales_by_Outlet_size.index, autopct='%.1f%%',startangle=90,colors=colors)

plt.title('Sales by Fat Content')
plt.axis('equal')
plt.show()
```

```python
# sales by Outlet Location
sales_by_location = df.groupby('Outlet Location Type')['Sales'].sum().reset_index()
sales_by_location = sales_by_location.sort_values('Sales',ascending=False)

plt.figure(figsize=(8,5))
ax = sns.barplot(x='Sales',y='Outlet Location Type',data=sales_by_location,color='#6ABA2E')



plt.title('Total Sales by Outlet Location')
plt.xlabel('Total Sales')
plt.ylabel('Outlet Location Type')

plt.tight_layout()
plt.show()
```
